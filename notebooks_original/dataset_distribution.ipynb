```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
df = pd.read_csv('shuffled_dataset_generalised/shuffled_metadata_generalized.csv')
# Display the first few rows of the dataset
df.head()
```

```python
print("✅ Total images:", len(df))
```

```python
# === 📊 1. Distribution by Ethnicity ===
ethnicity_counts = df["ethnicity"].value_counts().sort_values(ascending=False)
print("\nImages per ethnicity:\n", ethnicity_counts)

# === 📊 2. Distribution by Age Group ===
age_counts = df["age_group"].value_counts().sort_index()
print("\nImages per age group:\n", age_counts)

# === 📊 3. Distribution by Gender ===
gender_counts = df["gender"].value_counts()
print("\nImages per gender:\n", gender_counts)
```

```python
# === 📊 4. Cross-tab: Ethnicity x Age Group ===
ethnicity_age_matrix = pd.crosstab(df["ethnicity"], df["age_group"])
print("\nEthnicity x Age Group matrix:\n", ethnicity_age_matrix)
```

```python
# === 📊 5. Cross-tab: Ethnicity x Gender ===
ethnicity_gender_matrix = pd.crosstab(df["ethnicity"], df["gender"])
print("\nEthnicity x Gender matrix:\n", ethnicity_gender_matrix)
```

```python
# === 📊 6. Cross-tab: Age Group x Gender ===
age_gender_matrix = pd.crosstab(df["age_group"], df["gender"])
print("\nAge Group x Gender matrix:\n", age_gender_matrix)
```

```python
# 1. Bar plot: Ethnicity
ethnicity_counts.plot(kind="bar", figsize=(12, 4), title="Images per Ethnicity")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Bar plot: Age Group
age_counts.plot(kind="bar", figsize=(8, 4), color="skyblue", title="Images per Age Group")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# 3. Bar plot: Gender
gender_counts.plot(kind="bar", figsize=(6, 4), color="lightcoral", title="Images per Gender")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# 4. Heatmap: Ethnicity x Age Group
plt.figure(figsize=(12, 6))
sns.heatmap(ethnicity_age_matrix, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Heatmap: Ethnicity vs Age Group")
plt.ylabel("Ethnicity")
plt.xlabel("Age Group")
plt.tight_layout()
plt.show()
```

```python
# 5. Heatmap: Ethnicity x Gender
plt.figure(figsize=(10, 6))
sns.heatmap(ethnicity_gender_matrix, annot=True, fmt="d", cmap="YlOrRd")
plt.title("Heatmap: Ethnicity vs Gender")
plt.ylabel("Ethnicity")
plt.xlabel("Gender")
plt.tight_layout()
plt.show()
```

```python
# 6. Heatmap: Age Group x Gender
plt.figure(figsize=(8, 4))
sns.heatmap(age_gender_matrix, annot=True, fmt="d", cmap="Purples")
plt.title("Heatmap: Age Group vs Gender")
plt.ylabel("Age Group")
plt.xlabel("Gender")
plt.tight_layout()
plt.show()
```
